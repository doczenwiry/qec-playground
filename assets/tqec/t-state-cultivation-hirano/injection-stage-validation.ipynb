```python
import stim

circuit = stim.Circuit.from_file("stim/t-state-injection-surface-code.v4.compact.stim")

sim = stim.TableauSimulator()
sim.do(circuit)

xl = stim.PauliString("X0*X1*X2*X5*X6*X9*X10")
zl = stim.PauliString("Z0*Z1*Z2*Z5*Z6*Z9*Z10")

print("‹X_L›:", sim.peek_observable_expectation(xl))
print("‹Z_L›:", sim.peek_observable_expectation(zl))
```

```python
tableau = sim.current_inverse_tableau()**-1
for i, stab in enumerate(tableau.to_stabilizers()):
    print(i, stab)
```
