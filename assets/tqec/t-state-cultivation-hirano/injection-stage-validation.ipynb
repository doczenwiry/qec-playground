```python
import stim

circuit = stim.Circuit.from_file("stim/stage1-injection/t-state-injection-surface-code.v6.compact.stim")

sim = stim.TableauSimulator()
sim.do(circuit)

xl = stim.PauliString("X0*X1*X2*X6*X7*X10*X11")
zl = stim.PauliString("Z0*Z1*Z2*Z6*Z7*Z10*Z11")

print("‹X_L›:", sim.peek_observable_expectation(xl))
print("‹Z_L›:", sim.peek_observable_expectation(zl))
```

```python
tableau = sim.current_inverse_tableau()**-1
for i, stab in enumerate(tableau.to_stabilizers()):
    print(i, stab)
```
