```python
import stim

def compute_observable_expectation(circuit: stim.Circuit, qubits):
    simulator = stim.TableauSimulator()
    simulator.do(circuit)

    for observable in ["X", "Y", "Z"]:
        pauli = stim.PauliString("*".join(map(lambda q : observable + str(q), qubits)))
        print(f"<{observable}_L> : {simulator.peek_observable_expectation(pauli)}")

def count_cnots(circuit: stim.Circuit):
    count = 0
    for instruction in circuit.flattened():
        if instruction.name == "CX":
            count += len(instruction.targets_copy()) // 2
    print(f"Qubits: {circuit.num_qubits}")
    print(f"CNOTs : {count}")
```

```python
designs = {
    'layout1' : {
        'circuit' : "stim/stage1-injection/part1-steane-injection.layout1.v2.compact.stim",
        'qubits' : [0, 2, 3, 8, 9, 12, 14],
    },
    'layout2' : {
        'circuit' : "stim/stage1-injection/part1-steane-injection.layout2.v2.compact.stim",
        'qubits' : [0, 1, 5, 9, 10, 11, 15],
    }
}
```

```python
design = 'layout1'
circuit = stim.Circuit.from_file(designs[design]['circuit'])
qubits = designs[design]['qubits']
compute_observable_expectation(circuit, qubits)
count_cnots(circuit)
```

```python
design = 'layout2'
circuit = stim.Circuit.from_file(designs[design]['circuit'])
qubits = designs[design]['qubits']
compute_observable_expectation(circuit, qubits)
count_cnots(circuit)
```
