```python
import stim

circuit = stim.Circuit.from_file("stim/stage1-injection/t-state-injection-surface-code.v8.compact.stim")

sim = stim.TableauSimulator()
sim.do(circuit)

# For v6.compact layout
# qubits = [0, 1, 2, 6, 7, 10, 11]

# For v7.compact layout
# qubits = [0, 2, 3, 8, 9, 12, 14]

# For v8.compact layout
qubits = [0, 1, 5, 9, 10, 11, 15]

for obs in ["X", "Y", "Z"]:
    pauli = stim.PauliString("*".join(map(lambda q : obs + str(q), qubits)))
    print(f"<{obs}_L> : {sim.peek_observable_expectation(pauli)}")
```
