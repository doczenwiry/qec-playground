```python
import stim

designs = {
    'layout1' : {
        'circuit' : "stim/stage1-injection/t-state-injection-surface-code.layout1.v1.compact.stim",
        'qubits' : [0, 2, 3, 8, 9, 12, 14],
    },
    'layout2' : {
        'circuit' : "stim/stage1-injection/t-state-injection-surface-code.layout2.v1.compact.stim",
        'qubits' : [0, 1, 5, 9, 10, 11, 15],
    }
}
```

```python
def compute_observable_expectation(circuit, qubits):
    simulator = stim.TableauSimulator()
    simulator.do(circuit)

    for observable in ["X", "Y", "Z"]:
        pauli = stim.PauliString("*".join(map(lambda q : observable + str(q), qubits)))
        print(f"<{observable}_L> : {simulator.peek_observable_expectation(pauli)}")
```

```python
design = 'layout1'
compute_observable_expectation(
    circuit = stim.Circuit.from_file(designs[design]['circuit']),
    qubits = designs[design]['qubits']
)
```

```python
design = 'layout2'
compute_observable_expectation(
    circuit = stim.Circuit.from_file(designs[design]['circuit']),
    qubits = designs[design]['qubits']
)
```
