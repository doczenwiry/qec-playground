```python
import os
import stim
from tqecd import annotate_detectors_automatically

from notebooks.error_rate_analyser import noisify_phenomenological
```

```python
filename = "../assets/tqec/fixed_boundary_flipping-spatial-junction-x-basis-temporal-k=2.compact.cleared.stim.txt"
basename, _ = os.path.splitext(filename)
circuit = stim.Circuit().from_file(filename)
annotated = basename + ".annotated.stim"
annotate_detectors_automatically(circuit).to_file(annotated)
```

```python
circuit = stim.Circuit().from_file(annotated)

print(f"Missing detectors found by STIM: {len(circuit.missing_detectors())}")

noisy = noisify_phenomenological(circuit)
errors = noisy.shortest_graphlike_error(ignore_ungraphlike_errors=False, canonicalize_circuit_errors=True)

print(f"Length of shortest graph-like error : {len(errors)}")
for error in errors:
    print(error)

noisy.diagram("matchgraph-3d")
```
