```python
import itertools
import stim

from notebooks.circuit_expectations import count_cnots, compute_observable_expectation
from notebooks.circuit_flows import check_flow_preservation, check_syndrome_extraction, check_state_preparation
```

```python
circuit = stim.Circuit.from_file(
    "../assets/tqec/hirano-t-state-cultivation/stage1-injection/layout1/part1-modified-steane-code.layout1.v1.compact.stim"
)
logical = [0, 2, 3, 8, 9, 12, 14]
stabilizers = {
    'R' : [8, 9, 12, 14],
    'G' : [2, 3, 8, 12],
    'B' : [0, 3, 8, 9],
}

print("Stabilizer flows")
for stabilizer, support in itertools.product(['X', 'Z'], stabilizers.values()):
    check_state_preparation(circuit, stabilizer, support)

print(f"Observable flows")
check_state_preparation(circuit, 'Y', support=logical)
compute_observable_expectation(circuit, 'Y', logical)

print(f"Circuit statistics")
count_cnots(circuit)
```
