```python
from IPython.core.display import Markdown

from library.circuitry import Circuitry
from library.common import Pauli
from library.qubit_array import QubitArray
from utils.simulation.stim import sample
```

```python
scenarios_direct = {}
for observable in Pauli:
    qubits = QubitArray(dimensions=(1, 1))
    circuitry = Circuitry(qubits, clifford=True)

    circuitry.append("R", 0)
    circuitry.append_tick()
    circuitry.append("H", 0)
    circuitry.append_tick()
    circuitry.append("S", 0)
    circuitry.append_tick()
    circuitry.append("H", 0)
    circuitry.append_tick()
    circuitry.append("S", 0)
    circuitry.append_tick()

    circuitry.append_observable(0, "OBSERVABLE", {0: observable.name})

    scenario = rf"$\overline{{\mathbf{{{observable.name}}}}}$"
    scenarios_direct[scenario] = circuitry

    display(Markdown(f"[Open in Crumble ({scenario})]({circuitry.to_crumble_url()})"))
```

```python
scenarios_correct = {}
for observable in Pauli:
    qubits = QubitArray(dimensions=(3, 1))
    circuitry = Circuitry(qubits, clifford=True)

    circuitry.append("R", 0)
    circuitry.append("RX", [1, 2])
    circuitry.append("H", 0)
    circuitry.append("S", [1, 2])
    circuitry.append_tick()
    circuitry.append("CX", [1, 0])
    circuitry.append_tick()
    circuitry.append("CX", [0, 1])
    circuitry.append_tick()
    circuitry.append("CZ", [0, 1])
    circuitry.append_tick()
    circuitry.append("H", 1)
    circuitry.append_tick()
    circuitry.append("CX", [2, 1])
    circuitry.append_tick()
    circuitry.append("CX", [1, 2])
    circuitry.append_tick()
    circuitry.append("CZ", [1, 2])
    circuitry.append_tick()

    circuitry.append_observable(0, "OBSERVABLE", {2: observable.name})

    scenario = rf"$\overline{{\mathbf{{{observable.name}}}}}$"
    scenarios_correct[scenario] = circuitry

    display(Markdown(f"[Open in Crumble ({scenario})]({circuitry.to_crumble_url()})"))
```

```python
scenarios_tracked = {}
for observable in Pauli:
    qubits = QubitArray(dimensions=(3, 1))
    circuitry = Circuitry(qubits, clifford=True)

    circuitry.append("R", 0)
    circuitry.append("RX", [1, 2])
    circuitry.append_tick()
    circuitry.append("H", 0)
    circuitry.append("S", [1, 2])
    circuitry.append_tick()

    circuitry.append("CX", [1, 0])
    circuitry.append_tick()
    circuitry.append("H", 1)
    circuitry.append_tick()

    circuitry.append("CX", [2, 1])
    circuitry.append_tick()
    circuitry.append("MZ", 0)
    qubits.record_measurement(0, "MZ0")
    circuitry.append("MZ", 1)
    qubits.record_measurement(1, "MZ1")

    if observable == Pauli.Y:
        circuitry.append_observable(0, "OBSERVABLE", {2: observable.name})
    else:
        circuitry.append_observable(0, "OBSERVABLE", {2: observable.name}, "MZ1")

    scenario = rf"$\overline{{\mathbf{{{observable.name}}}}}$"
    scenarios_tracked[scenario] = circuitry

    display(Markdown(f"[Open in Crumble ({scenario})]({circuitry.to_crumble_url()})"))
```

```python
title = r"$\mathbf{HSHS}$ Experiment [Direct]"
sample(
    scenarios_direct,
    correction=True,
    title=title,
    label="Observable",
    fontsize=10,
    figsize=(16, 5),
)
```

```python
title = r"$\mathbf{HSHS}$ Experiment [Correct]"
sample(
    scenarios_correct,
    correction=True,
    title=title,
    label="Observable",
    fontsize=10,
    figsize=(16, 5),
)
```

```python
title = r"$\mathbf{HSHS}$ Experiment [Tracked]"
sample(
    scenarios_tracked,
    correction=True,
    title=title,
    label="Observable",
    fontsize=10,
    figsize=(16, 5),
)
```
