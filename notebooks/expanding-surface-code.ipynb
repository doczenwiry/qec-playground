```python
import stim
from IPython.core.display import Markdown

from library.qubit_array import QubitArray
from library.surface_code.expanding_patch import ExpandingSurfaceCodePatch
```

```python
circuit = stim.Circuit()
qubits = QubitArray(circuit, dimensions=(9,9))
expanding = ExpandingSurfaceCodePatch(qubits, distance=3, anchor=(1, 1), expansion=2)

for moment in ExpandingSurfaceCodePatch.MOMENTS:
    expanding.append_syndrome_slice(circuit, moment=moment, preparation=True, expanded=False)
    circuit.append("TICK")

for moment in ExpandingSurfaceCodePatch.MOMENTS:
    expanding.append_syndrome_slice(circuit, moment=moment, preparation=True, expanded=True)
    circuit.append("TICK")

display(Markdown(f"[Open in Crumble]({circuit.to_crumble_url()})"))
```

```python
print(circuit)
for polygon in expanding.get_polygons(expanded=True):
    print(polygon, end="")
```
