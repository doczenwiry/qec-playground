```python
from IPython.core.display import Markdown

from library.circuitry import Circuitry
from library.common import Pauli
from library.surface_code.patch import SurfaceCodePatch
from library.utils.faults.circuit_faults import analyse_heuristic
```

```python
DISTANCE = 5
qubits = SurfaceCodePatch.make_array(distance=DISTANCE)
circuitry = Circuitry(qubits=qubits)
basis = Pauli.Z

surface = SurfaceCodePatch(qubits, distance=DISTANCE)

circuitry.annotate_polygons(surface.get_polygons())

surface.append_memory(circuitry, prepare=Pauli.X, measure=Pauli.X, prefix="SFC:M0")

surface.annotate_detectors(
    circuitry, prepared=Pauli.X, measured=Pauli.X, prefix="SFC:M0:R"
)

last = DISTANCE - 1
circuitry.attach_observable(0, *(f"SFC:M0:R{last}:D{qi}" for qi in range(DISTANCE)))

circuitry.detectors_report()

label = f"Memory experiment [d={DISTANCE}]"
display(
    Markdown(f"[Open in Crumble ({label})]({circuitry.to_crumble_url(spacing=True)})")
)
```

```python
analyse_heuristic(circuitry.as_stim(per=0.01), label)
```
