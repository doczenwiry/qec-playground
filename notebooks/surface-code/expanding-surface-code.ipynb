```python
from IPython.core.display import Markdown

from library.circuitry import Circuitry
from library.qubit_array import QubitArray
from library.surface_code.expanding_patch import ExpandingSurfaceCodePatch
```

```python
qubits = QubitArray(dimensions=(9, 9))
circuitry = Circuitry(qubits)
expanding = ExpandingSurfaceCodePatch(qubits, distance=3, anchor=(1, 1), expansion=2)

circuitry.annotate_polygons(expanding.get_polygons(expanded=False))
for moment in ExpandingSurfaceCodePatch.MOMENTS:
    expanding.append_expansion(circuitry, moment=moment)
    circuitry.append_tick()
circuitry.annotate_polygons(expanding.get_polygons(expanded=True))

circuitry.to_file("../generated/expanding-surface-code", spacing=True)

display(Markdown(f"[Open in Crumble]({circuitry.to_crumble_url()})"))
```
