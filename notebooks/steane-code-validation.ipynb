```python
import itertools
import stim

from utils.circuit_expectations import (
    compute_stabilizer_expectation,
    compute_observable_expectation,
)
from utils.circuit_flows import check_syndrome_extraction, check_flow_preservation
from utils.simulation.noise import make_noisy_circuit
```

```python
circuit = stim.Circuit.from_file(
    "../assets/steane-code/steane-code-standard-syndrome-measurement.stim"
)
supports = {
    "R": [5, 6, 8, 9],
    "G": [0, 3, 5, 8],
    "B": [0, 1, 5, 6],
}

print("Stabilizer flows")
for stabilizer, support in itertools.product(["X", "Z"], supports.values()):
    check_flow_preservation(circuit, stabilizer, support)
print("Observable flows")
for observable in ["X", "Y", "Z"]:
    check_flow_preservation(circuit, observable, support=[0, 1, 3, 5, 6, 8, 9])
print("Syndrome extractions")
for syndrome, support in itertools.product(["X", "Z"], supports.values()):
    check_syndrome_extraction(circuit, syndrome, support)
```

```python
circuit = stim.Circuit.from_file(
    "../assets/steane-code/steane-code-standard-syndrome-measurement.annotated.stim"
)
supports = {
    "R": [5, 6, 8, 9],
    "G": [0, 3, 5, 8],
    "B": [0, 1, 5, 6],
}

noisy = make_noisy_circuit(circuit, per=0.01)

result = noisy.search_for_undetectable_logical_errors(
    dont_explore_detection_event_sets_with_size_above=4,
    dont_explore_edges_with_degree_above=3,
    dont_explore_edges_increasing_symptom_degree=False,
)

print(f"Minimal undetected logical error : {len(result)}")
for loc in result:
    print(loc)
```
