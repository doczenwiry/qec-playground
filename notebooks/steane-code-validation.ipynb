```python
import itertools
import stim

from notebooks.circuit_expectations import compute_stabilizer_expectation, compute_observable_expectation
from notebooks.circuit_flows import convert_to_pauli, extract_measurements
```

```python
design = stim.Circuit.from_file("../assets/steane-code/steane-code-standard-syndrome-measurement.stim")
supports = {
    'R' : [5, 6, 8, 9],
    'G' : [0, 3, 5, 8],
    'B' : [0, 1, 5, 6],
}
compute_stabilizer_expectation(design, supports)
compute_observable_expectation(design, observable='X', support=[0, 1, 3, 5, 6, 8, 9])
```

```python
circuit = stim.Circuit.from_file("../assets/steane-code/steane-code-superdense-syndrome-measurement.stim")
supports = {
    'R' : [6, 7, 11, 12],
    'G' : [0, 5, 6, 11],
    'B' : [0, 1, 6, 7],
}

measurements = extract_measurements(circuit) or list()
```

```python
print(f"Stabilizer flows")
for support, stabilizer in itertools.product(supports.values(), ['X', 'Z']):
    pauli = convert_to_pauli(stabilizer, support, qubits=circuit.num_qubits)
    solution = circuit.solve_flow_measurements([stim.Flow(input=pauli, output=pauli)])[0] or []
    flow = stim.Flow(input=pauli, output=pauli, measurements=solution)
    if circuit.has_flow(flow):
        report = f"Preserved (incl. meas. {list(map(lambda m: measurements[m], solution))})"
    else:
        report = f"DISTORTED"
    print(f"{stabilizer}({",".join(map(str, support))}) : {report}")
```

```python
print(f"Observable flows")
for support, observable in itertools.product([[0,1,5,6,7,11,12]], ['X', 'Y', 'Z']):
    pauli = convert_to_pauli(observable, support, qubits=circuit.num_qubits)
    solution = circuit.solve_flow_measurements([stim.Flow(input=pauli, output=pauli)])[0] or []
    flow = stim.Flow(input=pauli, output=pauli, measurements=solution)
    if circuit.has_flow(flow):
        report = f"Preserved (incl. meas. {list(map(lambda m: measurements[m], solution))})"
    else:
        report = f"DISTORTED"
    print(f"{observable}_L : {report}")
```

```python
print(f"Syndrome extractions")
identity = stim.PauliString(circuit.num_qubits)
for support, stabilizer in itertools.product(supports.values(), [ 'X', 'Z' ]):
    pauli = convert_to_pauli(stabilizer, support, qubits=circuit.num_qubits)
    solution = circuit.solve_flow_measurements([stim.Flow(input=pauli, output=identity)])[0] or []
    flow = stim.Flow(input=pauli, output=identity, measurements=solution)
    if circuit.has_flow(flow):
        report = f"Extracted (incl. meas. {list(map(lambda m: measurements[m], solution))})"
    else:
        report = f"DISTORTED"
    print(f"{stabilizer}({",".join(map(str, support))}) : {report}")
```
