```python
def detectors_report(circuit, open_boundaries: bool = False):
    print("Detector statistics : ")
    print(f"> Number of detectors : {circuit.num_detectors}")

    gauge_found = False
    try:
        circuit.detector_error_model(allow_gauge_detectors=False)
    except ValueError:
        gauge_found = True
    missing_detectors = circuit.missing_detectors(unknown_input=open_boundaries)
    print(f"> Missing detectors : {len(missing_detectors)}")
    for detector in missing_detectors:
        print(f">> Detector : {detector.targets_copy()}")
    print(f"> Gauge detectors : {'FOUND' if gauge_found else 'NONE'}")
```

```python
circuit = stim.Circuit.from_file("../../assets/steane-code/steane-code-standard-syndrome-measurement.observable.stim")
circuit += circuit.missing_detectors()

detectors_report(circuit)

analyse_heuristic(circuit)
```
