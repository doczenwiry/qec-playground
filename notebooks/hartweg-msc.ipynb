```python
from IPython.core.display import Markdown

from library.circuitry import Circuitry
from library.qubit_array import QubitArray
from library.surface_code.patch import SurfaceCodePatch
```

```python
SOURCE_DISTANCE = 3
TARGET_DISTANCE = 5

qubits: QubitArray = SurfaceCodePatch.make_array(TARGET_DISTANCE, (1, 1))
circuitry = Circuitry(qubits, clifford=True)
source = SurfaceCodePatch(qubits, distance=SOURCE_DISTANCE, anchor=(1, 1), coloring=False)
target = SurfaceCodePatch(qubits, distance=TARGET_DISTANCE, anchor=(1, 1), coloring=False)

circuitry.annotate_polygons(source.get_polygons(opacity=0.250))
circuitry.annotate_polygons(target.get_polygons(opacity=0.125))

display(Markdown(f"[Open in Crumble (Hartweg)]({circuitry.to_crumble_url()})"))
```
