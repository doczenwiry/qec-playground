```python
SOURCE_DISTANCE = 3
TARGET_DISTANCE = 5
scenarios = {}
```

```python
qubits: QubitArray = SurfaceCodePatch.make_array(TARGET_DISTANCE, (1, 1))
circuitry = Circuitry(qubits, clifford=True)
source = SurfaceCodePatch(qubits, distance=SOURCE_DISTANCE, anchor=(1, 1), coloring=False)
# target = SurfaceCodePatch(qubits, distance=TARGET_DISTANCE, anchor=(1, 1), coloring=False)

circuitry.annotate_polygons(source.get_polygons(opacity=0.250))
circuitry.append_tick()

source.append_injection(circuitry)

circuitry.append_observable(0, "OBSERVABLE_YL", source.logical(Pauli.Y))

scenario = "Injection"
scenarios[scenario] = circuitry

display(Markdown(f"[Open in Crumble ({scenario})]({circuitry.to_crumble_url()})"))
```

```python
qubits: QubitArray = SurfaceCodePatch.make_array(TARGET_DISTANCE, (1, 1))
circuitry = Circuitry(qubits, clifford=True)
source = SurfaceCodePatch(qubits, distance=SOURCE_DISTANCE, anchor=(1, 1), coloring=False)
# target = SurfaceCodePatch(qubits, distance=TARGET_DISTANCE, anchor=(1, 1), coloring=False)

circuitry.annotate_polygons(source.get_polygons(opacity=0.250))
circuitry.append_tick()

source.append_injection(circuitry)
source.append_checking(circuitry)

circuitry.attach_observable(0, "CHK:X")

scenario = "Checking"
scenarios[scenario] = circuitry

display(Markdown(f"[Open in Crumble ({scenario})]({circuitry.to_crumble_url()})"))
```

```python
sample(scenarios, label="Point")
```

```python
simulate(scenarios, label="Point", shots=1e7)
```
