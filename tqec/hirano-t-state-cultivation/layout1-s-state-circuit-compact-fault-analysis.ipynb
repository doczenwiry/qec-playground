# Fault analysis

## Imports and definitions

```python
import itertools

import stim
from IPython.core.display import Markdown

from library.circuitry import Circuitry
from library.common import Pauli
from library.qubit_array import QubitArray
from library.steane_code.compact_patch import SteaneCodePatch
from utils.circuit_faults import analyse_heuristic
```

```python
def append_mpp(circuitry: Circuitry, steane: SteaneCodePatch, stage: str, style: str):
    for label, stabilizer in steane.pauli_stabilizers(style).items():
        circuitry.append("MPP", stim.PauliString(stabilizer))
        qubits.record_measurement(f"{stage}_{label}")
        circuitry.append_tick()
    circuitry.append(
        "MPP",
        stim.PauliString(
            steane.logical(Pauli.Y, compact=style == "COMPACT", final=style == "FINAL")
        ),
    )
    qubits.record_measurement(f"{stage}_YL")
    circuitry.append_tick()
```

## Analysis of Injection stage

```python
qubits: QubitArray = QubitArray(dimensions=(6, 6))
circuitry = Circuitry(qubits, clifford=True)
steane = SteaneCodePatch(qubits=qubits, anchor=(1, 1))

circuitry.annotate_polygons(steane.get_polygons())

steane.append_injection(circuitry)
append_mpp(circuitry, steane, "MEASUREMENT", style="COMPACT")

# Annotate the X&Z Steane stabilizer detectors
for label in steane.pauli_stabilizers().keys():
    circuitry.annotate_detector(f"MEASUREMENT_{label}")

# Attach the Y_L observable
circuitry.attach_observable(0, "MEASUREMENT_YL")

display(Markdown(f"[Open in Crumble (Injection)]({circuitry.to_crumble_url()})"))

circuitry.detectors_report(open_boundaries=True)
```

```python
circuit = circuitry.as_stim(per=0.01, noiseless_measurement_ticks=7)
analyse_heuristic(circuit)
```

## Analysis of Superdense Code Cycle

```python
qubits: QubitArray = QubitArray(dimensions=(6, 6))
circuitry = Circuitry(qubits, clifford=True)
steane = SteaneCodePatch(qubits=qubits, anchor=(1, 1))

circuitry.annotate_polygons(steane.get_polygons())
append_mpp(circuitry, steane, "PREPARATION", style="COMPACT")

steane.append_superdense_cycle(circuitry, prefix="STN:SDC0")

circuitry.annotate_polygons(steane.get_polygons(compact=False))
append_mpp(circuitry, steane, "MEASUREMENT", style="DIFFUSE")

# Annotate the X&Z preparation detectors and the Z measurement detectors
for label in steane.pauli_stabilizers().keys():
    circuitry.annotate_detector(f"PREPARATION_{label}", f"STN:SDC0:{label}")
    if label[0] == "Z":
        circuitry.annotate_detector(f"MEASUREMENT_{label}", f"STN:SDC0:{label}")

# Annotate the X measurement detectors
circuitry.annotate_detector("MEASUREMENT_XR", "STN:SDC0:XG")
circuitry.annotate_detector("MEASUREMENT_XG", "STN:SDC0:XR")
circuitry.annotate_detector("MEASUREMENT_XB")

# Attach the Y_L observable
circuitry.attach_observable(
    0,
    "MEASUREMENT_YL",
    "PREPARATION_YL",
    *(f"STN:SDC0:{lb}" for lb in ["XR", "XG", "XB"]),
)

display(Markdown(f"[Open in Crumble (Superdense)]({circuitry.to_crumble_url()})"))

circuitry.detectors_report(open_boundaries=True)
```

```python
circuit = circuitry.as_stim(
    per=0.01, noiseless_preparation_ticks=7, noiseless_measurement_ticks=7
)
analyse_heuristic(circuit)
```

## Analysis of Cultivation stage

```python
qubits: QubitArray = QubitArray(dimensions=(6, 6))
circuitry = Circuitry(qubits, clifford=True)
steane = SteaneCodePatch(qubits=qubits, anchor=(1, 1))

circuitry.annotate_polygons(steane.get_polygons(compact=False))

append_mpp(circuitry, steane, "PREPARATION", style="DIFFUSE")
steane.append_doublecheck(circuitry)
append_mpp(circuitry, steane, "MEASUREMENT", style="DIFFUSE")

# Annotate the X&Z preparation&measurement detectors
for label in steane.pauli_stabilizers().keys():
    circuitry.annotate_detector(f"PREPARATION_{label}", f"MEASUREMENT_{label}")

# Annotate the X detectors
for label in ["RZ", "D1", "BZ", "EX0", "D4"]:
    circuitry.annotate_detector(f"STN:CULT:{label}")
for number in range(2):
    circuitry.annotate_detector(f"STN:CULT:Z{number}")

# Annotate the Y measurement detector
circuitry.annotate_detector("PREPARATION_YL", "STN:CULT:CC")
circuitry.annotate_detector("STN:CULT:CC", "STN:CULT:GZ")

# Attach the Y_L observable
circuitry.attach_observable(
    0, "MEASUREMENT_YL", *(f"STN:CULT:{lb}" for lb in ["D1", "D4", "RZ", "GZ"])
)

display(Markdown(f"[Open in Crumble (Cultivation)]({circuitry.to_crumble_url()})"))

circuitry.detectors_report(open_boundaries=True)
```

```python
circuit = circuitry.as_stim(
    per=0.01, noiseless_preparation_ticks=7, noiseless_measurement_ticks=7
)
analyse_heuristic(circuit)
```

## Analysis of Teleportation stage

```python
TELEPORTATION_ROUNDS = 3
qubits: QubitArray = QubitArray(dimensions=(6, 6))
circuitry = Circuitry(qubits, clifford=True)
steane = SteaneCodePatch(qubits=qubits, anchor=(1, 1))

circuitry.annotate_polygons(steane.get_polygons(compact=False))

append_mpp(circuitry, steane, "PREPARATION", style="DIFFUSE")
for rnd in range(TELEPORTATION_ROUNDS):
    steane.append_teleportation_round(circuitry, round=rnd, prefix=f"STN:TPT{rnd}")

circuitry.annotate_polygons(steane.get_polygons(final=True))

append_mpp(circuitry, steane, "MEASUREMENT", style="DIFFUSE")

# Annotate the X&Z preparation detectors and the Z measurement detectors
last = TELEPORTATION_ROUNDS - 1
for label in filter(lambda lb: lb != "XB", steane.pauli_stabilizers().keys()):
    circuitry.annotate_detector(f"PREPARATION_{label}", f"STN:TPT0:{label}")
    if label in ["ZR", "ZG", "ZB"]:
        circuitry.annotate_detector(f"MEASUREMENT_{label}", f"STN:TPT{last}:{label}")
    elif label == "XR":
        circuitry.annotate_detector(f"MEASUREMENT_{label}", f"STN:TPT{last}:XG")
    elif label == "XG":
        circuitry.annotate_detector(f"MEASUREMENT_{label}", f"STN:TPT{last}:XR")

# Annotate the XB detectors
# TODO: confirm whether Hirano does this as well
circuitry.annotate_detector("PREPARATION_XB", "MEASUREMENT_XB")
for rnd in range(TELEPORTATION_ROUNDS):
    circuitry.annotate_detector("PREPARATION_XB", "MEASUREMENT_XB", f"STN:TPT{rnd}:XB")

for label, (prev, curr) in itertools.product(
    steane.pauli_stabilizers().keys(), itertools.pairwise(range(TELEPORTATION_ROUNDS))
):
    match label[0]:
        case "X":
            if label == "XG":
                circuitry.annotate_detector(f"STN:TPT{curr}:XG", f"STN:TPT{prev}:XR")
            elif label == "XR":
                circuitry.annotate_detector(f"STN:TPT{curr}:XR", f"STN:TPT{prev}:XG")
        case "Z":
            circuitry.annotate_detector(
                f"STN:TPT{curr}:{label}", f"STN:TPT{prev}:{label}"
            )

# Attach the Y_L observable
circuitry.attach_observable(
    0,
    "PREPARATION_YL",
    "MEASUREMENT_YL",
    *(
        f"STN:TPT{rnd}:{ma}"
        for rnd, ma in itertools.product(
            range(TELEPORTATION_ROUNDS), ["XR", "XG", "XB"]
        )
    ),
)

display(Markdown(f"[Open in Crumble (Teleportation)]({circuitry.to_crumble_url()})"))

circuitry.detectors_report(open_boundaries=True)
```

```python
circuit = circuitry.as_stim(
    per=0.01, noiseless_preparation_ticks=7, noiseless_measurement_ticks=7
)
analyse_heuristic(circuit)
```
