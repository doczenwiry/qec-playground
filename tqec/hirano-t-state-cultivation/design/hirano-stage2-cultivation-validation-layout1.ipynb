```python
import stim

from utils.circuit_expectations import count_cnots
from utils.circuit_flows import check_flow_preservation, check_syndrome_extraction
```

```python
circuit = stim.Circuit.from_file(
    "stim/stage2-cultivation/layout1/part1-cultivation.layout1.v1.compact.stim"
)
logical = [0, 2, 3, 6, 7, 10, 11]
supports = {
    "R": [6, 7, 10, 11],
    "G": [2, 3, 6, 10],
    "B": [0, 3, 6, 7],
}

print("Stabilizer flows")
for support in supports.values():
    check_flow_preservation(circuit, "X", support, measurements=["MX 6"])
for support in supports.values():
    check_flow_preservation(circuit, "Z", support)
print("Observable flow")
check_flow_preservation(circuit, "Y", support=logical, measurements=["MX 6"])
print("Syndrome extraction")
check_syndrome_extraction(circuit, "Y", support=logical, measurements=["MX 6"])

print("Circuit statistics")
count_cnots(circuit)
```
