```python
import itertools
import stim

from utils.circuit_expectations import count_cnots, compute_observable_expectation
from utils.circuit_flows import (
    check_flow_preservation,
    check_syndrome_extraction,
    check_state_preparation,
)
```

```python
circuit = stim.Circuit.from_file(
    "stim/stage1-injection/layout2/part1-modified-steane-code.layout2.v1.compact.stim"
)
logical = [0, 1, 5, 9, 10, 11, 15]
stabilizers = {
    "R": [9, 10, 11, 15],
    "G": [0, 5, 9, 10],
    "B": [1, 5, 10, 11],
}

print("Stabilizer flows")
for stabilizer, support in itertools.product(["X", "Z"], stabilizers.values()):
    check_state_preparation(circuit, stabilizer, support)

print("Observable flows")
check_state_preparation(circuit, "Y", support=logical)
compute_observable_expectation(circuit, "Y", logical)

print("Circuit statistics")
count_cnots(circuit)
```
