```python
import itertools
import stim

from library.utils.circuit_expectations import (
    count_cnots,
    compute_observable_expectation,
)
from library.utils.circuit_flows import (
    check_flow_preservation,
    check_syndrome_extraction,
    check_state_preparation,
)
```

```python
circuit = stim.Circuit.from_file(
    "stim/stage1-injection/layout1/part1-modified-steane-code.layout1.v1.compact.stim"
)
logical = [0, 2, 3, 8, 9, 12, 14]
stabilizers = {
    "R": [8, 9, 12, 14],
    "G": [2, 3, 8, 12],
    "B": [0, 3, 8, 9],
}

print("Stabilizer flows")
for stabilizer, support in itertools.product(["X", "Z"], stabilizers.values()):
    check_state_preparation(circuit, stabilizer, support)

print("Observable flows")
check_state_preparation(circuit, "Y", support=logical)
compute_observable_expectation(circuit, "Y", logical)

print("Circuit statistics")
count_cnots(circuit)
```

```python
circuit = stim.Circuit.from_file(
    "stim/stage1-injection/layout1/part2-syndrome-superdense-cycle.layout1.bell.compact.stim"
)
logical = [0, 2, 3, 6, 7, 10, 11]
stabilizers = {
    "R": [6, 7, 10, 11],
    "G": [2, 3, 6, 10],
    "B": [0, 3, 6, 7],
}

print("Stabilizer flows")
for stabilizer, support in itertools.product(["X", "Z"], stabilizers.values()):
    check_flow_preservation(circuit, stabilizer, support)
print("Observable flows")
for observable in ["X", "Y", "Z"]:
    check_flow_preservation(circuit, observable, support=logical)
print("Syndrome extractions")
for syndrome, support in itertools.product(["X", "Z"], stabilizers.values()):
    check_syndrome_extraction(circuit, syndrome, support)

print("Circuit statistics")
count_cnots(circuit)
```
