```python
import stim

from library.utils.circuit_expectations import count_cnots
from library.utils.circuit_flows import (
    check_flow_preservation,
    check_syndrome_extraction,
)
```

```python
circuit = stim.Circuit.from_file("stim/gidney24/stage2-cultivation.gidney24.stim")
logical = [0, 3, 5, 7, 8, 10, 11]
supports = {
    "R": [0, 3, 5, 8],
    "G": [3, 5, 7, 10],
    "B": [5, 8, 10, 11],
}

print("Stabilizer flows")
for support in supports.values():
    check_flow_preservation(circuit, "X", support, measurements=["MX 5"])
for support in supports.values():
    check_flow_preservation(circuit, "Z", support)
print("Observable flows")
check_flow_preservation(circuit, "Y", support=logical, measurements=["MX 5"])
print("Syndrome extraction")
check_syndrome_extraction(circuit, "Y", logical, measurements=["MX 5"])

print("Circuit statistics")
count_cnots(circuit)
```
