```python
import itertools
import stim

from utils.circuit_expectations import count_cnots
from utils.circuit_flows import check_flow_preservation, check_syndrome_extraction
```

```python
circuit = stim.Circuit.from_file("stim/fowler/t-state-cultivation-fowler.stim")
logical = [0, 1, 4, 6, 8, 12, 13]
supports = {
    'R' : [6, 8, 12, 13],
    'G' : [0, 4, 6, 12],
    'B' : [0, 1, 6, 8],
}

print("Stabilizer flows")
for stabilizer, support in itertools.product(['X', 'Z'], supports.values()):
    check_flow_preservation(circuit, stabilizer, support)
print("Observable flows")
check_flow_preservation(circuit, 'Y', support=logical)
print("Syndrome extraction")
check_syndrome_extraction(circuit, 'Y', logical)

print(f"Circuit statistics")
count_cnots(circuit)
```
