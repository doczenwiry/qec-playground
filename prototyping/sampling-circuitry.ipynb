```python
from IPython.core.display import Markdown

from library.circuitry import Circuitry
from library.qubit_array import QubitArray
from utils.simulation.clifft import sample
```

```python
scenarios = dict()
for basis in ["X", "Y", "Z"]:
    qubits = QubitArray(dimensions=(1, 1))
    circuitry = Circuitry(qubits)
    circuitry.append("R", 0)
    circuitry.append("H", 0)
    circuitry.append("S", 0)
    circuitry.append("H", 0)
    circuitry.append("S", 0)
    circuitry.append(f"M{basis}", 0)
    circuitry.append_observable(0, "FINAL", {0: basis})
    scenarios[basis] = circuitry
    display(
        Markdown(
            f"[Open in Crumble ({basis}-basis)]({circuitry.as_stim.to_crumble_url()})"
        )
    )

sample(scenarios, title="HSHS experiment", label="Basis", shots=1e6)
```
